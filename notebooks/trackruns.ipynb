```python
import wandb
import pandas as pd
```

```python
from glob import glob
```

```python
api = wandb.Api()
```

```python
runids = glob("../model/checkpoints/*")
```

```python
variables_of_interest_model = ["n_fb", "fb_mode"]
variables_of_interest_env = [
    "T_iti",
    "reward_timing",
    "kernel_timing",
    "kappa_kernel",
    "xi_kernel",
]
variables_of_interest_decision = ["theta"]


def get_val(cfg, category, key):
    try:
        if category in cfg and key in cfg[category]:
            return cfg[category][key]
    except (TypeError, KeyError):
        pass
    flat_key = f"{category}.{key}"
    return cfg.get(flat_key, "N/A")


def extract_df(config):
    rows = []
    for k in variables_of_interest_decision:
        rows.append(
            {
                "Category": "Decision",
                "Parameter": k,
                "Value": get_val(config, "decision", k),
            }
        )

    for k in variables_of_interest_model:
        rows.append(
            {"Category": "Model", "Parameter": k, "Value": get_val(config, "model", k)}
        )

    for k in variables_of_interest_env:
        rows.append(
            {
                "Category": "Environment",
                "Parameter": k,
                "Value": get_val(config, "env", k),
            }
        )
    df = pd.DataFrame(rows)

    return df
```
